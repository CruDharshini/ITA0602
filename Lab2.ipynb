```python
import pandas as pd

data = {
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "AirTemp": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "High"],
    "Wind": ["Strong", "Strong", "Strong", "Strong"],
    "Water": ["Warm", "Warm", "Warm", "Cool"],
    "Forecast": ["Same", "Same", "Change", "Change"],
    "EnjoySport": ["Yes", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)
df.to_csv("trainingdata.csv", index=False)

print("trainingdata.csv created successfully!")
```

```python
import csv

# Read data from CSV
data = []
with open("trainingdata.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)  # Skip header
    for row in reader:
        data.append(row)

n = len(data[0]) - 1

# Initialize Specific and General hypotheses
S = ['0'] * n
G = ['?'] * n

for sample in data:
    if sample[-1] == "Yes":
        for i in range(n):
            if S[i] == '0':
                S[i] = sample[i]
            elif S[i] != sample[i]:
                S[i] = '?'

        for i in range(n):
            if S[i] != '?' and G[i] != S[i]:
                G[i] = S[i]

    else:
        for i in range(n):
            if S[i] != sample[i]:
                G[i] = S[i]
            else:
                G[i] = '?'

print("Specific Hypothesis (S):")
print(S)

print("\nGeneral Hypothesis (G):")
print(G)
```
